```python
import pandas as pd
import numpy as np

orders = {'o10001':{'date':'2024/01/10', 'product':'Hoodie', 'quantity':'1'},
            'o10002':{'date':'2024/01/13', 'product':'Tote bag', 'quantity':'2'},
            'o10003':{'date':'2024/01/14', 'product':'Pencil', 'quantity':'10'},
            'o10004':{'date':'2024/01/15', 'product':'T-shirt', 'quantity':'2'}
}

orders_df = pd.DataFrame(orders)
orders_df
```

```python
products = {'123':{'name':'Hoodie', 'cost_price':12.12, 'sale_price':15.00},
            '124':{'name':'Tote bag', 'cost_price':2.15, 'sale_price':9.99},
            '125':{'name':'Pencil', 'cost_price':22.45, 'sale_price':49.99},
            '126':{'name':'T-shirt', 'cost_price':0.45, 'sale_price':2.99},
            '127':{'name':'Honours degree', 'cost_price':0.78, 'sale_price':1.49}
}

products_df = pd.DataFrame(products)
products_df
```

```python
orders_df = orders_df.transpose()
products_df = products_df.transpose()
```

```python
orders_df
```

```python
# Ensure the DataFrames are properly transposed so that features like 'product' and 'name' are columns
orders_df_t = orders_df.T if 'product' not in orders_df.columns else orders_df
products_df_t = products_df.T if 'name' not in products_df.columns else products_df

# Perform the merge
joined_df = orders_df_t.merge(products_df_t, how='left', left_on='product', right_on='name')
joined_df
```

```python
joined_df = joined_df.drop(['name'], axis=1)
joined_df
```

```python
joined_df = orders_df_t.merge(
    products_df_t,
    how='left',
    left_on='product',
    right_on='name'
)
```

```python
joined_df = orders_df_t.merge(
    products_df_t,
    how='inner',
    left_on='product',
    right_on='name'
)

joined_df
```

```python
joined_df = orders_df_t.merge(
    products_df_t,
    how='right',
    left_on='product',
    right_on='name'
)

joined_df
```

```python
joined_df = orders_df_t.merge(
    products_df_t,
    how='outer',
    left_on='product',
    right_on='name'
)

joined_df
```

.merge() combines dataframes.



how= controls the type of join.



left_on= and right_on= specify which columns to match.
